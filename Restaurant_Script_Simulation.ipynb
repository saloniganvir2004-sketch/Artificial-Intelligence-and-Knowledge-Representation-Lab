# Restaurant Script Simulation Using Python

## 1. Introduction

In Artificial Intelligence, a **script** is a knowledge-representation structure used to describe a familiar, stereotyped sequence of events in a common situation. A restaurant visit is a useful example: customers generally enter, choose a table, read a menu, order food, eat, pay, and leave.

A script stores expectations about what normally happens, who participates, which objects are involved, and what conditions are expected before and after the event. This allows an AI system to interpret incomplete descriptions by using common, expected event sequences.

This notebook implements a simple restaurant script in Python. It retains the original nine-event sequence and develops it into a structured, reusable simulation.

To represent a typical restaurant visit as an AI script and simulate its sequence of events using Python.

## 5. Theory: Script-Based Knowledge Representation

A script represents a familiar situation as an organised sequence of events. It captures procedural or event-based knowledge about how a situation usually unfolds.

For example, when someone says, “Rahul had dinner at a restaurant,” a person can often infer that Rahul probably entered the restaurant, viewed a menu, ordered food, ate, paid, and left—even if every action is not explicitly stated.

### Main components of a script

| Component | Meaning | Restaurant example |
|---|---|---|
| Script name | Name of the familiar situation | Restaurant visit |
| Entry conditions | Conditions that should be true before the script begins | Customer is hungry and has money; restaurant is open |
| Roles | Participants and their responsibilities | Customer, waiter, cook, cashier |
| Props | Objects used during the situation | Table, menu, food, bill, money |
| Scenes | Ordered groups of events | Entering, ordering, eating, exiting |
| Track | A variation of the general script | Table-service restaurant, cafeteria, fast-food counter |
| Results | Conditions expected after the script | Customer has eaten and paid; customer has less money |

### Scenes and event order

A scene groups related actions. In this notebook, the restaurant visit is divided into four scenes:

1. **Entering:** The customer enters and gets seated.
2. **Ordering:** The menu is provided and the customer places an order.
3. **Eating:** Food is prepared, served, and consumed.
4. **Exiting:** The customer requests the bill, pays, and leaves.

The scenes preserve the original nine events from the supplied notebook while making their organisation clearer.

### Tracks

A track represents a variation of the general script. A table-service restaurant and a self-service cafeteria share the broad goal of obtaining food, but their event sequences may differ. This notebook demonstrates the standard table-service track only.


## 6. Importing Required Libraries

Only Python's built-in functionality is required. No external packages need to be installed.


```python
# No external libraries are required for this notebook.
print("Restaurant Script Simulation")
print("Using standard Python only.")
```

## 7. Defining the Restaurant Script

The script is stored as a Python dictionary. It contains the main components of a typical table-service restaurant visit. Each scene contains an ordered list of events.


```python
restaurant_script = {
    "name": "Restaurant Visit",
    "track": "Table-Service Restaurant",
    "entry_conditions": [
        "The customer is hungry.",
        "The customer has money to pay.",
        "The restaurant is open."
    ],
    "roles": {
        "customer": "Visits the restaurant, orders food, eats, and pays.",
        "waiter": "Provides the menu, takes the order, serves food, and presents the bill.",
        "cook": "Prepares the ordered food.",
        "cashier": "Receives payment."
    },
    "props": [
        "Restaurant", "Table", "Menu", "Food", "Bill", "Money"
    ],
    "scenes": {
        "Entering": [
            "Customer enters the restaurant",
        ],
        "Ordering": [
            "Waiter provides the menu",
            "Customer orders food"
        ],
        "Eating": [
            "Cook prepares food",
            "Waiter serves food",
            "Customer eats food"
        ],
        "Exiting": [
            "Customer requests the bill",
            "Customer pays the bill",
            "Customer leaves the restaurant"
        ]
    },
    "results": [
        "The customer has eaten.",
        "The customer has paid the bill.",
        "The customer has less money than before the visit."
    ]
}

print("Restaurant script created successfully.")
```

## 8. Displaying Script Components

The following code displays the script's metadata: entry conditions, roles, props, scenes, and expected results.


```python
print("SCRIPT:", restaurant_script["name"])
print("TRACK:", restaurant_script["track"])

print("\\nENTRY CONDITIONS:")
for condition in restaurant_script["entry_conditions"]:
    print("-", condition)

print("\\nROLES:")
for role, responsibility in restaurant_script["roles"].items():
    print(f"- {role.title()}: {responsibility}")

print("\\nPROPS:")
for prop in restaurant_script["props"]:
    print("-", prop)

print("\\nEXPECTED RESULTS:")
for result in restaurant_script["results"]:
    print("-", result)
```

## 9. Displaying Scenes in Sequence

A script is not merely a collection of unrelated events. The order of scenes and events represents the typical progression of the situation.


```python
print("RESTAURANT SCRIPT: EVENT SEQUENCE")
print("=" * 50)

event_number = 1
for scene_name, events in restaurant_script["scenes"].items():
    print(f"\\nSCENE: {scene_name.upper()}")
    for event in events:
        print(f"{event_number}. {event}")
        event_number += 1

print("\\n" + "=" * 50)
print("Restaurant script sequence displayed.")
```

## 10. Customer-Specific Simulation

The simulation function accepts a customer name and inserts it into the customer-related events. The remaining roles, such as the waiter and cook, remain unchanged.

This makes the same general script reusable for different customers.


```python
def restaurant_simulation(customer_name, script=restaurant_script):
    """Display a restaurant visit using the supplied customer's name."""

    if not isinstance(customer_name, str) or not customer_name.strip():
        raise ValueError("Customer name must be a non-empty string.")

    customer_name = customer_name.strip()
    print(f"RESTAURANT SIMULATION FOR {customer_name.upper()}")
    print("=" * 50)

    event_number = 1
    for scene_name, events in script["scenes"].items():
        print(f"\\nSCENE: {scene_name.upper()}")
        for event in events:
            personalised_event = event.replace("Customer", customer_name)
            print(f"{event_number}. {personalised_event}")
            event_number += 1

    print("\\nVISIT COMPLETED SUCCESSFULLY.")
    print("=" * 50)


restaurant_simulation("Rahul")
```

## 11. Simulating Another Customer

The function can be called again with a different name. The stored script does not need to be rewritten.


```python
restaurant_simulation("Ananya")
```

## 12. Script Validation

A simple validation function checks whether the script contains the expected components and whether every scene contains at least one event. This helps identify an incomplete script before simulation.


```python
def validate_restaurant_script(script):
    required_fields = [
        "name", "track", "entry_conditions",
        "roles", "props", "scenes", "results"
    ]

    missing_fields = [field for field in required_fields if field not in script]
    if missing_fields:
        return False, f"Missing required fields: {missing_fields}"

    if not script["scenes"]:
        return False, "The script must contain at least one scene."

    for scene_name, events in script["scenes"].items():
        if not isinstance(events, list) or len(events) == 0:
            return False, f"Scene '{scene_name}' must contain at least one event."

    if not script["roles"]:
        return False, "The script must define at least one role."

    if not script["props"]:
        return False, "The script must define at least one prop."

    return True, "All required script components are present."


is_valid, validation_message = validate_restaurant_script(restaurant_script)
print("Valid script:", is_valid)
print("Validation:", validation_message)
```

## 13. Automated Test Cases

The following tests verify that the script contains the expected nine events, that all four scenes are present, and that customer-specific simulation rejects an empty name.


```python
# Test 1: Verify the script name.
assert restaurant_script["name"] == "Restaurant Visit"

# Test 2: Verify all four scenes.
assert list(restaurant_script["scenes"].keys()) == [
    "Entering", "Ordering", "Eating", "Exiting"
]

# Test 3: Verify the original nine-event sequence is preserved.
total_events = sum(len(events) for events in restaurant_script["scenes"].values())
assert total_events == 9  # Original nine events.

# Test 4: Verify all expected roles are present.
assert {"customer", "waiter", "cook", "cashier"}.issubset(
    set(restaurant_script["roles"].keys())
)

# Test 5: Verify validation succeeds.
assert validate_restaurant_script(restaurant_script)[0] is True

# Test 6: Verify invalid customer input is rejected.
try:
    restaurant_simulation(" ")
except ValueError:
    invalid_name_rejected = True
else:
    invalid_name_rejected = False

assert invalid_name_rejected is True

print("All automated tests passed successfully.")
```
